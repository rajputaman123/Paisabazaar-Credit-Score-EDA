# **Project Summary -**

Paisabazaar helps customers find and apply for banking and credit products, and a core part of that
service depends on judging how creditworthy a customer is. This project works with a year's worth of
monthly financial snapshots (100,000 records covering 12,500 unique customers, 8 months each) collected
by Paisabazaar, capturing income, banking relationships, loan behaviour, payment habits and the resulting
Credit_Score label (Good / Standard / Poor).

The goal of this notebook is to explore that data end-to-end: understand what each column represents, check
its quality (missing values, duplicates, implausible values), and then build a structured, story-driven set
of visualisations that answer concrete questions Paisabazaar's credit team cares about — does income track
with credit score, does a poor credit mix or missed payments explain most of the "Poor" segment, does age or
occupation matter, and which handful of behavioural signals separate a "Good" customer from a "Poor" one.

The dataset arrives largely clean (no nulls or duplicate rows), which let the analysis focus on genuine
data-wrangling steps: turning delay counts, credit history length and utilisation ratios into interpretable
buckets, and re-expressing a few skewed monetary columns for cleaner visual comparison. The notebook first
profiles each variable on its own, then pairs the most meaningful predictors against Credit_Score, and
finally runs a couple of statistical tests to confirm those relationships aren't just visual noise.

The headline findings are intuitive but now quantified: customers with a "Good" credit score carry roughly
half the outstanding debt, are delayed on payments about a third as often, and hold nearly three fewer
credit cards/bank accounts on average than "Poor" customers. Credit_Mix and Payment_of_Min_Amount turn out
to be some of the strongest categorical signals, while Occupation barely moves the needle, which is itself a
useful (and slightly reassuring, from a fair-lending standpoint) negative finding.

These insights translate directly into business use for Paisabazaar: a small, explainable set of features —
payment delays, credit mix, number of open credit lines, outstanding debt and credit history length — that
a future credit-scoring model can lean on with confidence, and concrete, explainable talking points loan
officers can use when advising a customer on how to move from "Poor"/"Standard" into "Good" (pay down
revolving debt, avoid unnecessary new credit lines, pay above the minimum due). The notebook closes with a
short set of recommendations Paisabazaar's risk and product teams can act on directly.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# 1. Drop PII columns that carry no predictive/analytical value
df.drop(columns=['Name', 'SSN'], inplace=True)

# 2. Convert Credit_History_Age (in months) into years for easier interpretation
df['Credit_History_Age_Years'] = (df['Credit_History_Age'] / 12).round(1)

# 3. Bucket Age into readable groups for analysis
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[14, 25, 35, 45, 56],
    labels=['15-25', '26-35', '36-45', '46-56']
)

# 4. Clean up Type_of_Loan: mark customers with no recorded loan type explicitly
df['Type_of_Loan'] = df['Type_of_Loan'].replace('No Data', 'No Loan')

# 5. Fix category order for the target so charts read Poor -> Standard -> Good consistently
df['Credit_Score'] = pd.Categorical(df['Credit_Score'], categories=['Poor', 'Standard', 'Good'], ordered=True)

print("Wrangling complete. New shape:", df.shape)
df.head()
````

### What all manipulations have you done and insights you found?

Manipulations performed:
- Dropped Name and SSN — identifiers with no analytical value and a privacy liability if retained.
- Derived Credit_History_Age_Years from the month-based Credit_History_Age for more intuitive charts.
- Bucketed Age into four Age_Group bins (15-25, 26-35, 36-45, 46-56) for cleaner categorical comparisons.
- Relabelled the "No Data" placeholder in Type_of_Loan to "No Loan" for clarity.
- Fixed an explicit Poor -> Standard -> Good ordering on Credit_Score so every chart reads consistently.

Early insight: roughly 1 in 9 customers (11,408 of 100,000 records) have no loan on file at all, which is
worth flagging as a distinct profile from "has loans but pays on time."

## Hypothesis & Statistical Testing

Based on the charts above, one clear pattern is that customers with a 'Bad' Credit_Mix are far more likely
to have a 'Poor' Credit_Score. We test this statistically below.

Null Hypothesis (H0): Credit_Mix and Credit_Score are independent (no real relationship).
Alternate Hypothesis (H1): Credit_Mix and Credit_Score are dependent (related to each other).

````python
from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(df['Credit_Mix'], df['Credit_Score'])
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square statistic: {chi2:.2f}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Result: Reject the null hypothesis — Credit_Mix and Credit_Score are significantly related.")
else:
    print("Result: Fail to reject the null hypothesis.")
````

Since the p-value is far below 0.05, we reject the null hypothesis. This statistically confirms what the
visualizations already suggested — Credit_Mix has a real, significant relationship with Credit_Score, not
just a coincidental pattern in this sample.
